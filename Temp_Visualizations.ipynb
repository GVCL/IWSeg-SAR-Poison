## Corruption Visualizations for DeepLabV3, SegFormer, and Mask2Former

```python
import matplotlib.pyplot as plt
import os
from PIL import Image as PILImage
```

```python
def plotAllResults(
    corruptions,
    image_dir,
    corrupted_mask_parent_dir,
    predicted_mask_parent_dir,
    output_dir,
    model_type,
    number=-1
):
    mask_dir = os.path.join(predicted_mask_parent_dir, "0", "Outputs")
    mask_files = sorted(os.listdir(mask_dir))

    plt.rcParams.update({'font.size': 9})

    for idx, mask_file in enumerate(mask_files):
        
        print(idx)

        i_str = mask_file.split('_')[1]
        image_path = os.path.join(image_dir, i_str)
        arbitrary_img = PILImage.open(image_path).convert("L")

        fig, axes = plt.subplots(
            nrows=len(corruptions),
            ncols=3,
            figsize=(7, 2.3 * len(corruptions))
        )

        for row, corruption in enumerate(corruptions):
            
            print(corruption)

            corrupted_mask_dir = f'{corrupted_mask_parent_dir}/train_{corruption}_small'
            predicted_mask_dir = f'{predicted_mask_parent_dir}/{corruption}/Outputs'

            corrupt_mask_path = os.path.join(corrupted_mask_dir, i_str)
            predicted_mask_path = os.path.join(predicted_mask_dir, f"pred_{i_str}")

            corrupt_mask = PILImage.open(corrupt_mask_path).convert("L")
            predicted_mask = PILImage.open(predicted_mask_path).convert("L")

            # ---- Column 1: Input image
            axes[row, 0].imshow(arbitrary_img, cmap="gray")
            # ---- Column 2: Corrupted GT
            axes[row, 1].imshow(corrupt_mask, cmap="gray", vmin=0, vmax=255)
            # ---- Column 3: Prediction
            axes[row, 2].imshow(predicted_mask, cmap="gray", vmin=0, vmax=255)
            
            subtitle_text = f'{corruption}% corruption'

            axes[row, 0].text(
                -0.05, 0.5,                     # x < 0 puts it outside the axis
                subtitle_text,
                verticalalignment='center',
                horizontalalignment='right',
                rotation=90,
                transform=axes[row, 0].transAxes,
                fontsize=9,
                color='black'
            )

            for col in range(3):
                axes[row, col].axis("off")

        # ---- Column headers (only once)
        column_titles = [
            "(a) Input Image",
            "(b) Corrupted Ground Truth",
            "(c) Predicted Mask"
        ]

        for col, title in enumerate(column_titles):
            axes[0, col].set_title(title, fontsize=10, pad=10)
            
        plt.subplots_adjust(
            left=0.12,
            right=0.98,
            top=0.92,
            bottom=0.05,
            wspace=0.02,   # ← column spacing (key)
            hspace=0.02    # ← row spacing
        )

        os.makedirs(output_dir, exist_ok=True)

        # ---- High-resolution save (THIS IS WHAT REVIEWERS WANT)
        plt.savefig(
            os.path.join(output_dir, f"{i_str}"),
            dpi=600,
            bbox_inches="tight"
        )
        plt.close()

        if idx + 1 == number:
            break
```

```python
corruptions = [0,2,12,15,17,20,25,30]
models = ["DeepLabOutputs", "Segformer", "Maskformer"]

for model in models:
    
    image_dir = "./sar_images/images/train"
    corrupted_mask_parent_dir = "./sar_images/masks"
    predicted_mask_parent_dir = f"./GEE_Output/{model}/Train"
    output_dir = f"./GEE_Output/Train_Visualizations/{model}"

    plotAllResults(corruptions, image_dir, corrupted_mask_parent_dir, predicted_mask_parent_dir, output_dir, model)
```

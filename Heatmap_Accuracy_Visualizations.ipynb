```python
import os
import cv2
from skimage.metrics import structural_similarity as ssim
import tifffile as tiff
import glob
import re

def find_max_ssim(reference_img_path, folder_path):
    ref = cv2.imread(reference_img_path, cv2.IMREAD_GRAYSCALE)

    max_ssim = -1
    best_image = None

    for fname in os.listdir(folder_path):
        img_path = os.path.join(folder_path, fname)

        img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
        if img is None:
            continue

        # Resize if shapes differ
        if img.shape != ref.shape:
            img = cv2.resize(img, (ref.shape[1], ref.shape[0]))

        score = ssim(ref, img)

        if score > max_ssim:
            max_ssim = score
            best_image = fname

    return best_image, max_ssim
```

```python
import rasterio
import numpy as np

def tif_to_png(tif_path, png_path):
    with rasterio.open(tif_path) as src:
        img = src.read(1)  # first band

    # Normalize to 0–255
    img = img.astype(np.float32)
    img = 255 * (img - img.min()) / (img.max() - img.min() + 1e-8)
    img = img.astype(np.uint8)

    cv2.imwrite(png_path, img)

# Example
tif_to_png("./data_new/29.tif", "output.png")
```
